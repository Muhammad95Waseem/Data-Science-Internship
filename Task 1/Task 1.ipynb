```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data_set = pd.read_csv('iris.csv')
```

```python
data_set.shape
```

```python
data_set.columns
```

```python
data_set.head()
```

```python
plt.scatter(data_set['SepalLengthCm'], data_set['PetalLengthCm'])
plt.title('Sepal vs. Petal Lenght')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Langht')
plt.show()
```

```python
plt.scatter(data_set['SepalWidthCm'], data_set['PetalWidthCm'])
plt.title('Sepal vs. Petal Width')
plt.xlabel('Sepal Width')
plt.ylabel('Petal width')
plt.show()
```

```python
plt.figure(figsize=(10, 10))
sns.histplot(data=data_set, x='SepalLengthCm', bins=20, color='green')
plt.title('SepalLength Distribution')
plt.show()
```

```python
plt.figure(figsize=(10, 10))
sns.histplot(data=data_set, x='PetalLengthCm', bins=20, color='green')
plt.title('PetalLength Distribution')
plt.show()
```

```python
plt.figure(figsize=(10, 10))
sns.boxplot(data=data_set, x='SepalLengthCm', color='purple')
plt.title('Spread of Sepal Lenght')
```

```python
plt.figure(figsize=(10, 10))
sns.boxplot(data=data_set, x='PetalLengthCm', color='red')
plt.title('Spread of Petal Lenght')
```
